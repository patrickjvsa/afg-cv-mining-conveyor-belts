Esta plantilla consulta métricas persistidas; no define una evaluación científica. Primero prepare y monte Drive mediante `00_setup.ipynb` en esta máquina. Ajuste la ruta de sesión si cambió work_dir.

En Colab puede ejecutar este paso desde la última sección de `00_setup.ipynb` con `run_step("02_evaluacion.ipynb")`; así usa la misma sesión y configuración.


```python
from pathlib import Path
import json
SESSION_FILE = Path(globals().get("SESSION_FILE", "/content/cv-work/session.json"))
if not SESSION_FILE.is_file(): raise FileNotFoundError("Ejecute 00_setup en esta máquina.")
session = json.loads(SESSION_FILE.read_text())
import sys
source_dir = str(Path(session["repo_dir"]) / "src")
if source_dir not in sys.path: sys.path.insert(0, source_dir)
from cv_project.experiments import collect_runs
records = collect_runs(session["paths"]["results"])
print("Experimentos finalizados:", len(records))
for record in records: print(json.dumps(record, ensure_ascii=False, indent=2))
```
